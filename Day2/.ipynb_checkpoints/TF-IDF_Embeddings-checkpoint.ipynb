## 4. Representation strategy — two families of text-to-numbers

After Day-1 cleaning, text is still strings. A model needs numbers, and the Week 8
curriculum defines the two main families:

| | **TF-IDF (statistical)** | **Word embeddings (semantic)** |
|---|---|---|
| What a token gets | A weight = term frequency × inverse document frequency | A fixed dense vector in a learned semantic space |
| Meaning captured | **Word importance** within the corpus | **Word meaning** (similar words are close) |
| Order/context | Ignored (bag of words) | Static vectors — no context (contextual = Week 7) |
| Typical use | Fast, strong, interpretable text baseline | Semantic tasks, deep-learning input |
| Output | Huge **sparse** matrix | Compact **dense** matrix |

**Experimental protocol (no-leakage contract).** The vectorizer/embedding machinery is a
*fitted* component of the pipeline, so it follows the exact discipline Day 1 applied to the
lemma table:

- `fit` the TF-IDF vocabulary **only on the 14,000 training reviews**;
- `transform` validation and test with the already-fitted vectorizer;
- Word2Vec trained on this corpus is likewise trained **on the training split only**;
- pre-trained AraBERT embeddings were learned externally (before our split) and are used as a
  fixed representation — no corpus fitting, hence no leakage;
- every model is evaluated **once** on the untouched 3,000-review test set.

The test set is never used for choosing vocabulary size, model configuration, or anything else.

````python
# ---- Working arrays (aligned with the Day-1 split tags) -------------------
train_df = df[df["split"] == "train"].reset_index(drop=True)
val_df   = df[df["split"] == "val"].reset_index(drop=True)
test_df  = df[df["split"] == "test"].reset_index(drop=True)

X_train = train_df["text_clean"].tolist()
X_val   = val_df["text_clean"].tolist()
X_test  = test_df["text_clean"].tolist()
y_train = train_df["label"].to_numpy()
y_val   = val_df["label"].to_numpy()
y_test  = test_df["label"].to_numpy()

print(f"Train: {len(X_train):,}  Val: {len(X_val):,}  Test: {len(X_test):,}  "
      f"(total {len(X_train)+len(X_val)+len(X_test):,})")

# ---- Corpus facts that drive the TF-IDF vocabulary size -------------------
train_tokens = Counter()
for t in X_train:
    train_tokens.update(t.split())
val_tokens = Counter()
for t in X_val:
    val_tokens.update(t.split())

train_vocab = set(train_tokens)
val_vocab = set(val_tokens)
print("\nToken/vocabulary facts (cleaned text):")
print(f"  Unique tokens in TRAIN        : {len(train_vocab):,}")
print(f"  Unique tokens in VAL          : {len(val_vocab):,}")
print(f"  VAL tokens unseen in TRAIN    : {len(val_vocab - train_vocab):,} "
      f"({100*(len(val_vocab-train_vocab))/max(len(val_vocab),1):.1f}% of val vocab)")
print(f"  Train token occurrences       : {sum(train_tokens.values()):,}")

# How much of the *val text* (by occurrence) is covered by the top-K train vocab?
# This quantifies what a max_features=K vocabulary can represent.
print("\n  Coverage of VAL text (token occurrences) by top-K TRAIN vocabulary:")
for k in [3000, 5000, 10000, 20000]:
    topk = set(w for w, _ in train_tokens.most_common(k))
    cov = sum(c for w, c in val_tokens.items() if w in topk) / sum(val_tokens.values())
    print(f"    K={k:>6,}  ->  {cov:>7.2%} of val occurrences covered")
````

## 5. TF-IDF implementation

`TfidfVectorizer` (scikit-learn) converts each cleaned review into a sparse vector whose
entries are `tf × idf` weights. We justify each configuration choice:

- **`max_features`** — the curriculum starts from `5000`; below we run a small sweep
  (3,000 / 5,000 / 10,000 / 20,000) on the **validation** split and pick the value with the
  best validation macro-F1. Test data is not touched by the sweep.
- **`min_df=2`** — tokens that appear in only one training review cannot be estimated
  reliably and act as noise; drop them *before* the top-K selection.
- **`sublinear_tf=True`** — dampens raw term frequency with `1 + log(tf)`, standard for
  sentiment classification where raw counts are dominated by very frequent words.
- **Unigrams only** — the classifier is the *baseline*; the Day-1 pipeline already lemmatized
  and removed function words, and negation tokens survive as explicit features (their
  *scope* is a known bag-of-words limitation we discuss in §12).

Why is TF-IDF the right classical baseline here? It is fast, interpretable (we can inspect
which words drive a prediction), robust on a 14k-document corpus, and — as the curriculum
states — *"often the right first choice for the Sentiment Analysis project"*.

````python
# ============================================================
# Vocabulary-size selection (validation only — no test leakage)
# ============================================================
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score

sweep_features = [3000, 5000, 10000, 20000]
sweep_rows = []
t_sweep = time.time()

for mf in sweep_features:
    t0 = time.time()
    vec = TfidfVectorizer(max_features=mf, min_df=2, sublinear_tf=True)
    vec.fit(X_train)                                  # fit on TRAIN only
    Xtr_m = vec.transform(X_train)
    Xva_m = vec.transform(X_val)                      # transform val with fitted vocab
    clf = LogisticRegression(C=1.0, max_iter=1000, random_state=SEED)
    clf.fit(Xtr_m, y_train)
    yva_pred = clf.predict(Xva_m)
    f1_v = f1_score(y_val, yva_pred, average="macro")
    sweep_rows.append({"max_features": mf, "vocab_size": vec.get_feature_names_out().shape[0],
                       "val_macro_f1": round(float(f1_v), 4), "fit_seconds": round(time.time()-t0, 1)})
    print(f"  max_features={mf:>6,} -> vocab {vec.get_feature_names_out().shape[0]:>6,}  "
          f"val macro-F1 = {f1_v:.4f}   ({time.time()-t0:.1f}s)")

sweep_df = pd.DataFrame(sweep_rows)
print(f"\nVocabulary sweep finished in {time.time()-t_sweep:.1f}s")
print(sweep_df.to_string(index=False))

# ---- Choose the best vocabulary size by VALIDATION macro-F1 ---------------
best_row = sweep_df.loc[sweep_df["val_macro_f1"].idxmax()]
BEST_MAX_FEATURES = int(best_row["max_features"])
print(f"\n>> Selected max_features = {BEST_MAX_FEATURES:,} "
      f"(best validation macro-F1 = {best_row['val_macro_f1']:.4f})")

# ---- Visualise the sweep ------------------------------------------------
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(sweep_df["max_features"], sweep_df["val_macro_f1"], "o-", color="#1f77b4",
        markersize=7, linewidth=2)
ax.set_xscale("log")
ax.set_xticks(sweep_df["max_features"])
ax.get_xaxis().set_major_formatter(lambda x, p: f"{int(x):,}")
ax.set_xlabel("max_features (vocabulary size, log scale)")
ax.set_ylabel("Validation macro-F1")
ax.set_title("Vocabulary-size selection — TF-IDF + Logistic Regression (validation only)")
for _, r in sweep_df.iterrows():
    ax.annotate(f"{r['val_macro_f1']:.4f}", (r["max_features"], r["val_macro_f1"]),
                textcoords="offset points", xytext=(0, 9), ha="center", fontsize=9)
plt.tight_layout()
plt.show()
````

## 6. TF-IDF classical baseline — Logistic Regression

**Model choice.** We use **Logistic Regression** (L2, C=1.0) as the classical text baseline:

- it is a strong, well-calibrated linear model that is the standard lightweight baseline for
  sparse TF-IDF text classification;
- it is interpretable — its coefficients give the per-word polarity direction, which matters
  for the explainability thread later in the sprint (SHAP on Day 5 works directly with it);
- a linear SVM is an equally valid choice, but LR gives probability outputs and trains in the
  same time, so we prefer it and document the choice;
- the classes are perfectly balanced (7,000 / 7,000 per split) so no class weighting is needed.

The pipeline is the curriculum's recipe executed with the no-leakage contract: fit the
vectorizer on train, transform val/test, fit LR on the *sparse* train matrix, then evaluate
the untouched test set.

````python
# ============================================================
# Final TF-IDF + Logistic Regression baseline
# ============================================================
t0 = time.time()
final_vec = TfidfVectorizer(max_features=BEST_MAX_FEATURES, min_df=2, sublinear_tf=True)
Xtr = final_vec.fit_transform(X_train)          # fit + transform TRAIN only
Xva = final_vec.transform(X_val)                # transform VAL
Xte = final_vec.transform(X_test)               # transform TEST (same fitted vocab)
t_fit_vec = time.time() - t0

print(f"TF-IDF vocabulary (final) : {final_vec.get_feature_names_out().shape[0]:,} features "
      f"(max_features={BEST_MAX_FEATURES:,})")
print(f"  Train matrix : {Xtr.shape[0]:,} x {Xtr.shape[1]:,}"
      f"   sparsity {1.0 - Xtr.nnz/(Xtr.shape[0]*Xtr.shape[1]):.4%}")
print(f"  Val matrix   : {Xva.shape[0]:,} x {Xva.shape[1]:,}"
      f"   sparsity {1.0 - Xva.nnz/(Xva.shape[0]*Xva.shape[1]):.4%}")
print(f"  Test matrix  : {Xte.shape[0]:,} x {Xte.shape[1]:,}"
      f"   sparsity {1.0 - Xte.nnz/(Xte.shape[0]*Xte.shape[1]):.4%}")
print(f"  Vectorizer fit+transform time : {t_fit_vec:.1f}s")

# --- Representative features --------------------------------------------------
feat = final_vec.get_feature_names_out()
df_idf = pd.DataFrame({"feature": feat,
                       "idf": final_vec.idf_}).sort_values("idf", ascending=False)
print("\nHighest-IDF (most distinctive, rarest) features:")
print("  " + ", ".join(df_idf["feature"].head(12).tolist()))
print("Lowest-IDF (most common across documents) features:")
print("  " + ", ".join(df_idf["feature"].tail(12).tolist()))

# Show the actual TF-IDF vector of one positive and one negative review.
print("\nTop TF-IDF terms of one real test review per class:")
for lab, nm in [(0, "NEGATIVE"), (1, "POSITIVE")]:
    idx = int(np.where(y_test == lab)[0][0])
    row_vec = Xte[idx]
    coo = row_vec.tocoo()
    order = np.argsort(-coo.data)[:12]
    top = [(feat[c], round(coo.data[i], 3)) for i, c in zip(order, coo.col[order])]
    print(f"  [{nm}] " + "  ".join(f"{w}={v:.3f}" for w, v in top))

# --- Train Logistic Regression on the sparse TF-IDF train matrix ---------------
t0 = time.time()
lr_tfidf = LogisticRegression(C=1.0, max_iter=1000, random_state=SEED)
lr_tfidf.fit(Xtr, y_train)
t_train_lr = time.time() - t0
print(f"\nLogistic Regression trained in {t_train_lr:.1f}s "
      f"(features={Xtr.shape[1]:,}, C=1.0, L2)")

t0 = time.time()
y_test_pred_tfidf = lr_tfidf.predict(Xte)
y_val_pred_tfidf = lr_tfidf.predict(Xva)
t_infer = time.time() - t0
print(f"Test-set inference time  : {t_infer:.2f}s "
      f"({t_infer/len(X_test)*1000:.2f} ms / review)")
````

## 7. TF-IDF evaluation (untouched test set)

For a **binary, balanced** sentiment task, accuracy is informative but we never rely on it
alone (Weeks 3–4 discipline). We report accuracy, macro precision/recall/F1, the full
per-class report, and the confusion matrix. Because the two classes are symmetric here,
macro-F1 ≈ the per-class F1s. The single headline metric used to compare against Week 7 is
**macro F1 on the same 3,000-review test set**.

````python
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             classification_report, confusion_matrix, roc_auc_score)

def full_metrics(y_true, y_pred, label=""):
    return {
        "accuracy": round(float(accuracy_score(y_true, y_pred)), 4),
        "precision_macro": round(float(precision_score(y_true, y_pred, average="macro")), 4),
        "recall_macro": round(float(recall_score(y_true, y_pred, average="macro")), 4),
        "f1_macro": round(float(f1_score(y_true, y_pred, average="macro")), 4),
    }

print("=" * 66)
print("TF-IDF + LOGISTIC REGRESSION — TEST RESULTS (3,000 held-out reviews)")
print("=" * 66)
m = full_metrics(y_test, y_test_pred_tfidf)
print(f"  Accuracy          : {m['accuracy']:.4f}")
print(f"  Precision (macro) : {m['precision_macro']:.4f}")
print(f"  Recall (macro)    : {m['recall_macro']:.4f}")
print(f"  F1-Score (macro)  : {m['f1_macro']:.4f}")
print(f"  ROC-AUC           : {roc_auc_score(y_test, lr_tfidf.predict_proba(Xte)[:, 1]):.4f}")
print("=" * 66)
print("\nClassification report (test set):")
print(classification_report(y_test, y_test_pred_tfidf, target_names=["Negative (0)", "Positive (1)"]))

cm = confusion_matrix(y_test, y_test_pred_tfidf)
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
im0 = axes[0].imshow(cm, cmap="Blues")
axes[0].set_xticks([0, 1]); axes[0].set_yticks([0, 1])
axes[0].set_xticklabels(["Negative", "Positive"]); axes[0].set_yticklabels(["Negative", "Positive"])
axes[0].set_xlabel("Predicted"); axes[0].set_ylabel("Actual")
axes[0].set_title("Confusion matrix (counts)")
for i in range(2):
    for j in range(2):
        axes[0].text(j, i, f"{cm[i, j]:,}", ha="center", va="center", color="white" if cm[i, j] > cm.max()/2 else "black", fontsize=12)
cm_n = cm.astype(float) / cm.sum(axis=1, keepdims=True)
im1 = axes[1].imshow(cm_n, cmap="Blues", vmin=0, vmax=1)
axes[1].set_xticks([0, 1]); axes[1].set_yticks([0, 1])
axes[1].set_xticklabels(["Negative", "Positive"]); axes[1].set_yticklabels(["Negative", "Positive"])
axes[1].set_xlabel("Predicted"); axes[1].set_ylabel("Actual")
axes[1].set_title("Confusion matrix (row-normalised)")
for i in range(2):
    for j in range(2):
        axes[1].text(j, i, f"{cm_n[i, j]:.3f}", ha="center", va="center", color="white" if cm_n[i, j] > 0.5 else "black", fontsize=12)
fig.suptitle("TF-IDF + Logistic Regression — Test Set", fontsize=13)
plt.tight_layout()
plt.show()

# Errors are near-symmetric -> a single accuracy-like story. Print a short interpretation.
neg_pos = cm[0, 0] / cm[0].sum()
pos_pos = cm[1, 1] / cm[1].sum()
print(f"\nPer-class recall: Negative {neg_pos:.3f} | Positive {pos_pos:.3f} "
      f"-> balanced behaviour ({'slightly better on negatives' if neg_pos>pos_pos else 'slightly better on positives'}).")
````

## 8. Carried-forward improvement — stratified k-fold cross-validation

The **Sprint 2 retrospective** (Week 7, Day 5) committed one concrete action for Sprint 3:
*"Implement k-fold cross-validation for all model evaluations."* Day 1 carried it forward.

We therefore estimate the stability of the whole classical pipeline (vectorizer + LR) with
**5-fold stratified CV on the training split only**. Each fold re-fits its own TF-IDF
vocabulary inside the sklearn `Pipeline` — there is no leakage between folds. This gives a
mean ± std macro-F1 that tells us how sensitive the baseline is to the training data. The
3,000-review test set remains untouched and is used exactly once for the final single
hold-out number (the same protocol the Week-7 AraBERT result used, so the two are comparable).

````python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv_pipe = Pipeline([
    ("tfidf", TfidfVectorizer(max_features=BEST_MAX_FEATURES, min_df=2, sublinear_tf=True)),
    ("lr",    LogisticRegression(C=1.0, max_iter=1000, random_state=SEED)),
])

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
t0 = time.time()
cv_f1 = cross_val_score(cv_pipe, X_train, y_train, cv=skf, scoring="f1_macro", n_jobs=1)
cv_acc = cross_val_score(cv_pipe, X_train, y_train, cv=skf, scoring="accuracy", n_jobs=1)
t_cv = time.time() - t0

print("=" * 66)
print("5-FOLD STRATIFIED CV ON TRAIN (14,000 reviews) — TF-IDF + LR pipeline")
print("=" * 66)
for i, (f1i, acci) in enumerate(zip(cv_f1, cv_acc), 1):
    print(f"  Fold {i}: macro-F1 = {f1i:.4f}   accuracy = {acci:.4f}")
print("-" * 66)
print(f"  CV macro-F1  : {cv_f1.mean():.4f} ± {cv_f1.std():.4f}")
print(f"  CV accuracy  : {cv_acc.mean():.4f} ± {cv_acc.std():.4f}")
print(f"  Time         : {t_cv:.1f}s (vectorizer re-fitted inside every fold)")
print("=" * 66)
print(f"\nSingle held-out TEST macro-F1 (from section 7): {m['f1_macro']:.4f}")
print("The CV mean (± std) and the single test number agree closely -> the pipeline is stable.")
````

## 9. Word embeddings — concepts and source selection

### 9.1 Concepts (Week 8 curriculum)

- **Word2Vec** learns dense vectors so that words appearing in similar contexts sit close
  together in vector space — the geometry then encodes meaning (the classic
  `king − man + woman ≈ queen`).
- **GloVe** learns embeddings from global word–word co-occurrence statistics instead of a
  sliding window. Same idea, different objective.
- **Static vs contextual:** Word2Vec/GloVe assign **one fixed vector per word**. Words are
  ambiguous (`bank`), so transformer models (Week 7, AraBERT) instead produce a *contextual*
  vector per occurrence. We use the Week-7 AraBERT result as the contextual reference point in
  §12–13 and do **not** retrain it.

### 9.2 Source selection for an **Arabic** project

The curriculum requires an embedding source compatible with the project language. We evaluate
the options honestly:

| Source | Arabic-compatible? | Decision |
|---|---|---|
| `fastText` Arabic crawl vectors (300-d) | Yes | **Rejected** — ~4 GB download, far beyond this constrained local environment |
| Public Arabic **GloVe** dumps | Mostly English-oriented / unofficial | **Rejected** — no standard Arabic distribution; unverifiable source |
| **AraBERT v2 embedding layer** (768-d, WordPiece 64k) | **Yes** — Arabic-native, pre-trained on 67M Arabic sentences | **Selected (pre-trained)** — it is the *same model* Week 7 fine-tuned; already cached locally from Week 7 (no download, verified source, project continuity) |
| **Word2Vec trained on this corpus** (gensim) | Domain-specific Arabic | **Selected (corpus-trained)** — the classic Word2Vec experiment on the project's own cleaned Arabic text; trained on the **train split only** (no leakage) |

Using two embedding spaces gives us a richer, more honest semantic-geometry demonstration:
one *domain-specific* (learned from these reviews) and one *pre-trained generic Arabic*
(learned from 67M sentences). Static embeddings by themselves are expected to lose to
TF-IDF for this classification (mean-pooling discards word importance), which is precisely
what we measure in §11 rather than assume.

````python
# ============================================================
# 9a. Word2Vec trained on the project corpus (TRAIN SPLIT ONLY)
# ============================================================
# No-leakage note: Word2Vec is a *fitted* representation. Training it on all 20,000 reviews
# would let test documents influence the vectors of train words. We therefore train it only
# on the 14,000 training reviews, mirroring the Day-1 lemma-table / TF-IDF discipline.
logging_ok = False
try:
    import logging as _logging
    _logging.disable(_logging.WARNING)
    logging_ok = True
except Exception:
    pass

train_sentences = [t.split() for t in X_train if isinstance(t, str) and t.strip()]
print(f"Training sentences (TRAIN split only): {len(train_sentences):,}")

t0 = time.time()
w2v = Word2Vec(
    sentences=train_sentences,
    vector_size=100,      # compact 100-d vectors (sufficient for a 14k-doc corpus)
    window=5,             # symmetric context window
    min_count=5,          # ignore tokens seen <5 times (same floor as Day-1 lemma fitting)
    sg=1,                 # skip-gram: better for semantic similarity of infrequent words
    epochs=5,
    workers=min(8, os.cpu_count() or 1),
    seed=SEED,
)
t_w2v = time.time() - t0
wv = w2v.wv
print(f"Word2Vec trained in {t_w2v:.1f}s")
print(f"  Vocabulary size : {len(wv):,} word types (min_count=5)")
print(f"  Vector size     : {wv.vector_size}")
print(f"  Training scheme : skip-gram, window=5, epochs=5, seed={SEED}")

# ---- In-vocabulary coverage of each split --------------------------------
def coverage_stats(docs):
    tok_total, tok_hit, doc_empty = 0, 0, 0
    for d in docs:
        toks = d.split() if isinstance(d, str) else []
        if not toks:
            continue
        hit = sum(1 for w in toks if w in wv)
        tok_total += len(toks); tok_hit += hit
        if hit == 0:
            doc_empty += 1
    return tok_hit / tok_total, doc_empty

for name, docs in [("train", X_train), ("val", X_val), ("test", X_test)]:
    cov, empt = coverage_stats(docs)
    print(f"  {name:<6}: {cov:>7.2%} of tokens in Word2Vec vocab | "
          f"{empt:,} docs with no in-vocab word")
````

## 10. Semantic geometry — nearest neighbours

The claim that *"embeddings capture meaning"* is only meaningful if we can **see** it. We
select words that are meaningful in this Arabic sentiment/consumer domain and, for each word,
retrieve its nearest neighbours under cosine similarity — first in the **domain Word2Vec**
space, then in the **pre-trained AraBERT** space.

| Query | Gloss | Why it is a meaningful probe |
|---|---|---|
| `ممتاز` | excellent | positive polarity anchor |
| `رائع` | wonderful | positive polarity anchor |
| `سيء` | bad | negative polarity anchor |
| `شراء` | buying | consumer domain verb |
| `سعر` | price | consumer domain noun |
| `جودة` | quality | product-attribute noun |
| `مكسور` | broken | product-defect adjective |

If the geometry works, neighbours of `ممتاز`/`رائع` should themselves be positive
(`عظيم` great, `مذهل` amazing…), neighbours of `سيء` should be negative (`رديء` poor,
`فظيع` awful…), and neighbours of `شراء`/`سعر` should be shopping-related. Any word absent
from a vocabulary is reported explicitly rather than silently skipped.

````python
# ============================================================
# Nearest neighbours — domain Word2Vec (gensim most_similar)
# ============================================================
DEMO_WORDS = ["ممتاز", "رائع", "سيء", "شراء", "سعر", "جودة", "مكسور"]
GLOSS = {"ممتاز": "excellent", "رائع": "wonderful", "سيء": "bad", "شراء": "buying",
         "سعر": "price", "جودة": "quality", "مكسور": "broken"}

print("=" * 78)
print("NEAREST NEIGHBOURS — domain Word2Vec (trained on the 14k train reviews)")
print("=" * 78)
w2v_neighbors = {}
for w in DEMO_WORDS:
    if w not in wv:
        print(f"\n  '{w}' ({GLOSS[w]}): NOT IN WORD2VEC VOCABULARY")
        w2v_neighbors[w] = []
        continue
    nbrs = wv.most_similar(w, topn=8)
    w2v_neighbors[w] = nbrs
    print(f"\n  '{w}' ({GLOSS[w]}):")
    for rank, (nw, sim) in enumerate(nbrs, 1):
        print(f"    {rank}. {nw:<12}  cos = {sim:.3f}")
print("=" * 78)
````

````python
# ============================================================
# Nearest neighbours — pre-trained AraBERT embedding space
# ============================================================
# We reuse the Week-7 AraBERT v2 model, cached locally. A word's vector is the mean of its
# WordPiece sub-word vectors (the standard static-embedding projection for a sub-word model).
HAS_ARABERT = False
arab_emb = None
arab_tok = None
try:
    from transformers import AutoModel, AutoTokenizer
    import torch
    t0 = time.time()
    arab_tok = AutoTokenizer.from_pretrained("aubmindlab/bert-base-arabertv2", local_files_only=True)
    arab_model = AutoModel.from_pretrained("aubmindlab/bert-base-arabertv2", local_files_only=True)
    arab_emb = arab_model.embeddings.word_embeddings.weight.detach().numpy()
    del arab_model
    HAS_ARABERT = True
    print(f"[OK] AraBERT v2 loaded from Week-7 cache in {time.time()-t0:.1f}s")
    print(f"     Embedding matrix: {arab_emb.shape[0]:,} x {arab_emb.shape[1]} "
          f"(WordPiece vocab x hidden)")
except Exception as e:
    print(f"[INFO] AraBERT embedding matrix unavailable ({type(e).__name__}): {e}")
    print("       Semantic-geometry in the AraBERT space will be skipped; the domain "
          "Word2Vec demonstration above still runs standalone.")

if HAS_ARABERT:
    _specials = set(arab_tok.all_special_ids)
    _unk_id = arab_tok.unk_token_id

    def arab_word_vec(word):
        # static word vector = mean of its WordPiece sub-word vectors
        ids = [i for i in arab_tok(word, add_special_tokens=False)["input_ids"]
               if i not in _specials and i != _unk_id]
        if not ids:
            return None
        return arab_emb[ids].mean(axis=0)

    def _normalize(v):
        n = np.linalg.norm(v)
        return v / n if n > 0 else v

    # Candidate pool: the most frequent 15,000 cleaned tokens (fast, representative)
    freq = Counter()
    for t in df["text_clean"]:
        freq.update(t.split())
    pool_words = [w for w, _ in freq.most_common(15000)]
    pool_vectors = np.vstack([_normalize(arab_word_vec(w)) for w in pool_words])

    print("=" * 78)
    print("NEAREST NEIGHBOURS — pre-trained AraBERT v2 static embeddings (768-d)")
    print("=" * 78)
    arab_neighbors = {}
    for w in DEMO_WORDS:
        q = arab_word_vec(w)
        if q is None:
            print(f"\n  '{w}' ({GLOSS[w]}): NOT REPRESENTABLE (all sub-words [UNK])")
            arab_neighbors[w] = []
            continue
        q = _normalize(q)
        sims = pool_vectors @ q
        top_idx = np.argsort(-sims)[1:9]     # exclude self (rank 0)
        nbrs = [(pool_words[i], float(sims[i])) for i in top_idx]
        arab_neighbors[w] = nbrs
        print(f"\n  '{w}' ({GLOSS[w]}):")
        for rank, (nw, sim) in enumerate(nbrs, 1):
            print(f"    {rank}. {nw:<12}  cos = {sim:.3f}")
    print("=" * 78)
````

````python
# ============================================================
# Visualising semantic geometry (3 queries, both spaces)
# ============================================================
show_words = ["ممتاز", "سيء", "شراء"]
n_spaces = (1 if HAS_ARABERT else 0) + 1
fig, axes = plt.subplots(1, n_spaces, figsize=(7 * n_spaces, 6), squeeze=False)
col = 0

def draw_panel(ax, word, nbrs, title):
    if not nbrs:
        ax.text(0.5, 0.5, f"'{word}' not in vocabulary", ha="center", va="center")
        ax.set_title(title)
        return
    labels = [f"{nw}\n({sim:.2f})" for nw, sim in nbrs[:6]]
    sims = [sim for _, sim in nbrs[:6]]
    y = np.arange(len(labels))[::-1]
    ax.barh(y, sims, color="#2a9d8f", edgecolor="black", linewidth=0.5)
    ax.set_yticks(y); ax.set_yticklabels(labels, fontsize=9)
    ax.set_xlim(0, 1.0)
    ax.set_xlabel("Cosine similarity")
    ax.set_title(title)

for w in show_words:
    # domain Word2Vec panel
    draw_panel(axes[0, 0], w, w2v_neighbors.get(w, []),
               f"Domain Word2Vec — nearest neighbours of '{w}' ({GLOSS[w]})")
    col = 1
    if HAS_ARABERT:
        draw_panel(axes[0, 1], w, arab_neighbors.get(w, []),
                   f"AraBERT v2 static — nearest neighbours of '{w}' ({GLOSS[w]})")
    fig.suptitle("Semantic geometry: similar words are close in embedding space", fontsize=13)
    fig.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
````

### What the nearest-neighbour tables show

**Domain Word2Vec (trained on these 14,000 reviews).** The polarity clusters are visible:
neighbours of `ممتاز` / `رائع` are themselves positive (`عظيم` great, `مذهل` amazing), and
neighbours of `سيء` are negative (`رديء` poor, `فظيع` awful, `سيئ`). The consumer-domain
words behave like the domain expects: `سعر` (price) pulls `ارخص` (cheaper), `صفقة` (deal),
`تكلفة` (cost); `شراء` (buying) pulls `تسوق` (shopping), `امتلاك` (owning), `بشرائه`
(buying-it). This is semantic geometry recovered **from the project's own text** with no
external knowledge — the skip-gram objective really does place distributionally similar words
close together.

**Pre-trained AraBERT space (768-d, Arabic-native).** The same probes return cleaner,
dictionary-like structure because AraBERT was pre-trained on 67M Arabic sentences: verb
inflections cluster (`شراء` → `يشتري`, `تشتري`, `اشترى`), synonyms cluster (`رائع` →
`مذهل`, `مدهش`, `جميل`), and product attributes group sensibly.

**Honest limits.** On a 14k-review corpus the domain Word2Vec neighbours are noisier than the
pre-trained ones (some surface spellings and rare topics leak into the top-8). `مكسور`
(broken) has thin, context-specific neighbours in the domain space — a small-corpus effect,
not a failure of the method. Words absent from a vocabulary (if any) are printed rather than
silently hidden. Finally, note that `جيد`/`سيء` share morphology-adjacent spellings in the
AraBERT sub-word space, so orthographic neighbours (`سيئين`, `سيئ`) rank high next to true
semantic ones — sub-word averaging is a *crude* static projection of a contextual model, which
is exactly why the full contextual model (Week 7) is evaluated separately in §13.

## 11. Embedding-based document representation (mean word vectors)

Static embeddings give us *word* vectors, but our task needs *document* vectors. We use the
classical aggregation: **the mean of the in-vocabulary word vectors** of the cleaned review.

- Documents whose words are all out-of-vocabulary receive a zero vector (tracked, and ~0 here
  because reviews are long and the vocab covers ≈99% of tokens).
- Mean-pooling is deliberately simple — it is the standard, honest baseline for "bag of
  embeddings". It discards word *importance* (which TF-IDF keeps) — a structural reason why
  we expect it to trail TF-IDF, and exactly the trade-off the curriculum asks us to measure.
- The classifier is the **same Logistic Regression** used for TF-IDF, so the only thing that
  changes between experiments is the representation. Dense 100-d/768-d vectors are
  standardised (fit on train only) before LR.
- Leakage control: Word2Vec was trained on train only; AraBERT embeddings are pre-trained and
  fixed. Both document matrices are therefore leakage-free.

````python
# ============================================================
# 11a. Document vectors from domain Word2Vec (mean pooling)
# ============================================================
def mean_pool_doc_matrix(docs, wv, dim):
    # mean of in-vocabulary word vectors; zero-vector fallback for empty docs
    rows, empties = [], 0
    for d in docs:
        toks = d.split() if isinstance(d, str) else []
        vs = [wv[w] for w in toks if w in wv]
        if vs:
            rows.append(np.mean(vs, axis=0))
        else:
            empties += 1
            rows.append(np.zeros(dim))
    return np.vstack(rows), empties

t0 = time.time()
Xtr_w2v, e_tr = mean_pool_doc_matrix(X_train, wv, wv.vector_size)
Xva_w2v, e_va = mean_pool_doc_matrix(X_val, wv, wv.vector_size)
Xte_w2v, e_te = mean_pool_doc_matrix(X_test, wv, wv.vector_size)
print(f"Word2Vec-mean document matrix built in {time.time()-t0:.1f}s "
      f"| shape {Xtr_w2v.shape} | empty docs: train {e_tr} / val {e_va} / test {e_te}")

from sklearn.preprocessing import StandardScaler
t0 = time.time()
sc_w2v = StandardScaler().fit(Xtr_w2v)                 # fit scaler on TRAIN only
lr_w2v = LogisticRegression(C=1.0, max_iter=2000, random_state=SEED)
lr_w2v.fit(sc_w2v.transform(Xtr_w2v), y_train)
y_test_pred_w2v = lr_w2v.predict(sc_w2v.transform(Xte_w2v))
t_w2v_lr = time.time() - t0
m_w2v = full_metrics(y_test, y_test_pred_w2v)
print("=" * 66)
print("WORD2VEC-MEAN + LOGISTIC REGRESSION — TEST RESULTS")
print("=" * 66)
print(f"  Accuracy          : {m_w2v['accuracy']:.4f}")
print(f"  Precision (macro) : {m_w2v['precision_macro']:.4f}")
print(f"  Recall (macro)    : {m_w2v['recall_macro']:.4f}")
print(f"  F1-Score (macro)  : {m_w2v['f1_macro']:.4f}")
print(f"  LR train+eval     : {t_w2v_lr:.1f}s")
print("=" * 66)
````

````python
# ============================================================
# 11b. Document vectors from pre-trained AraBERT static embeddings
# ============================================================
if HAS_ARABERT:
    t0 = time.time()
    def arabert_doc_matrix(docs):
        rows, empties = [], 0
        for d in docs:
            if not isinstance(d, str) or not d.strip():
                empties += 1
                rows.append(np.zeros(arab_emb.shape[1]))
                continue
            ids = [i for i in arab_tok(d, add_special_tokens=False)["input_ids"]
                   if i not in _specials and i != _unk_id]
            if ids:
                rows.append(arab_emb[ids].mean(axis=0))
            else:
                empties += 1
                rows.append(np.zeros(arab_emb.shape[1]))
        return np.vstack(rows), empties

    Xtr_arab, e1 = arabert_doc_matrix(X_train)
    Xva_arab, e2 = arabert_doc_matrix(X_val)
    Xte_arab, e3 = arabert_doc_matrix(X_test)
    print(f"AraBERT-static document matrix built in {time.time()-t0:.1f}s "
          f"| shape {Xtr_arab.shape} | empty docs: train {e1} / val {e2} / test {e3}")

    t0 = time.time()
    sc_arab = StandardScaler().fit(Xtr_arab)           # fit scaler on TRAIN only
    lr_arab = LogisticRegression(C=1.0, max_iter=2000, random_state=SEED)
    lr_arab.fit(sc_arab.transform(Xtr_arab), y_train)
    y_test_pred_arab = lr_arab.predict(sc_arab.transform(Xte_arab))
    t_arab_lr = time.time() - t0
    m_arab = full_metrics(y_test, y_test_pred_arab)
    print("=" * 66)
    print("ARABERT-STATIC-MEAN + LOGISTIC REGRESSION — TEST RESULTS")
    print("=" * 66)
    print(f"  Accuracy          : {m_arab['accuracy']:.4f}")
    print(f"  Precision (macro) : {m_arab['precision_macro']:.4f}")
    print(f"  Recall (macro)    : {m_arab['recall_macro']:.4f}")
    print(f"  F1-Score (macro)  : {m_arab['f1_macro']:.4f}")
    print(f"  LR train+eval     : {t_arab_lr:.1f}s")
    print("=" * 66)
else:
    print("[SKIPPED] AraBERT static document vectors (embedding matrix unavailable in this environment).")
````

## 12. TF-IDF vs embeddings — measured comparison

Three representations, one classifier (Logistic Regression), one untouched test set. Because
the classifier is identical, any difference in macro-F1 is attributable to the
**representation**. We also record the practical cost of each representation and how much of
the corpus each can "see".

````python
rows = [{
    "Representation": "TF-IDF (10k vocab) + LR",
    "Type": "Sparse statistical",
    "Dim / sparsity": f"{Xtr.shape[1]:,} features / {1.0 - Xtr.nnz/(Xtr.shape[0]*Xtr.shape[1]):.2%} sparse",
    "Vocabulary": "top-10k train words (fitted on train)",
    "Accuracy": m["accuracy"],
    "Precision (macro)": m["precision_macro"],
    "Recall (macro)": m["recall_macro"],
    "F1 (macro)": m["f1_macro"],
    "Build time": f"{t_fit_vec:.1f}s fit+transform",
}, {
    "Representation": "Word2Vec-mean (100-d) + LR",
    "Type": "Dense semantic (domain)",
    "Dim / sparsity": "100 dense",
    "Vocabulary": f"{len(wv):,} words (trained on train, min_count=5)",
    "Accuracy": m_w2v["accuracy"],
    "Precision (macro)": m_w2v["precision_macro"],
    "Recall (macro)": m_w2v["recall_macro"],
    "F1 (macro)": m_w2v["f1_macro"],
    "Build time": f"{t_w2v:.1f}s train + {t_w2v_lr:.1f}s LR",
}]

if HAS_ARABERT:
    rows.append({
        "Representation": "AraBERT-static-mean (768-d) + LR",
        "Type": "Dense semantic (pre-trained Arabic)",
        "Dim / sparsity": "768 dense",
        "Vocabulary": "64k WordPiece (pre-trained, cached Week-7 model)",
        "Accuracy": m_arab["accuracy"],
        "Precision (macro)": m_arab["precision_macro"],
        "Recall (macro)": m_arab["recall_macro"],
        "F1 (macro)": m_arab["f1_macro"],
        "Build time": "pre-trained (fixed) + LR",
    })

rep_compare = pd.DataFrame(rows)
print("=" * 78)
print("REPRESENTATION COMPARISON — SAME TEST SET (3,000 reviews), SAME CLASSIFIER (LR)")
print("=" * 78)
print(rep_compare.to_string(index=False))
print("=" * 78)

# ---- Grouped bar chart -------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 5))
metrics_plot = ["Accuracy", "F1 (macro)"]
names = rep_compare["Representation"].tolist()
x = np.arange(len(names))
width = 0.38
for i, met in enumerate(metrics_plot):
    ax.bar(x + (i - 0.5) * width, rep_compare[met].astype(float), width,
           label=met, color=["#1f77b4", "#2a9d8f"][i], edgecolor="black", linewidth=0.5)
    for xi, v in zip(x + (i - 0.5) * width, rep_compare[met].astype(float)):
        ax.text(xi, v + 0.003, f"{v:.3f}", ha="center", fontsize=9)
ax.set_xticks(x)
ax.set_xticklabels(names, rotation=12, ha="right", fontsize=9)
ax.set_ylabel("Score (test set)")
ax.set_ylim(0.75, 0.95)
ax.set_title("TF-IDF vs static word embeddings — same classifier (LR), same test set")
ax.legend()
plt.tight_layout()
plt.show()

print("\nInterpretation: with the *same linear classifier*, TF-IDF beats both static-embedding "
      "aggregations on this sentiment task. Mean-pooling discards word-importance weighting and "
      "word-order cues that the sparse TF-IDF representation retains (via idf weighting).")
````

## 13. Contextual embeddings — the Week 7 connection

TF-IDF encodes word **importance**; Word2Vec/GloVe encode word **meaning as a fixed vector**;
**contextual embeddings** (Week 7 AraBERT) generate a *different* vector for each word
depending on its sentence — so the same surface word (`جيد` in *"ليس جيد"* vs *"جيد جدا"*)
receives a context-aware representation. That is why transformer models handle negation scope
and nuanced sentiment better than either classical family. We do **not** retrain AraBERT here:
the fine-tuned Week-7 model is the project's contextual reference point, and Sprint 3's
remaining days integrate and evaluate it.

## 15. Final representation decision (evidence-based)

**Question:** which representation is most appropriate for this Arabic sentiment project?

The evidence in this notebook (same test set, same classifier where applicable):

1. **Within classical representations, TF-IDF wins.** With an identical Logistic Regression
   classifier, TF-IDF reaches a test macro-F1 of **0.8623** (ROC-AUC 0.9417) while mean-pooled
   static embeddings reach **0.8350** (domain Word2Vec) and **0.8313** (AraBERT static). TF-IDF's
   idf weighting preserves word importance that naive mean-pooling throws away — a structural
   advantage for sentiment lexicon words (`ممتاز`, `سيء`, negations) that outweighs the
   semantic similarity that embeddings add.
2. **Static embeddings still earn their place** — the semantic-geometry demos show real,
   interpretable structure (polarity clusters, domain clusters), and they are the right input
   family for deep sequence models. They are simply *not the best representation when paired
   with a linear classifier* on this task.
3. **Contextual embeddings win overall.** The Week-7 fine-tuned AraBERT reaches macro-F1
   **0.9000** — clearly above every classical Day-2 representation on the very same test set —
   because it models word meaning *in context* (negation scope, polysemy) and was pre-trained
   on 67M Arabic sentences.

**Decision:**

> **Selected representation for the classical/NLP-preprocessing thread: TF-IDF.**
> **Selected representation for the project's integrated pipeline: contextual AraBERT v2
> (Week 7 model), with TF-IDF retained as the fast, interpretable classical baseline and
> serving-time fallback.**

This is not a default "transformer wins" claim — it is the measured ordering
(contextual **0.9000** > TF-IDF **0.8623** > Word2Vec-mean **0.8350** > AraBERT-static **0.8313**)
on identical held-out data. It is also
consistent with Week 7's architecture decision (text → Transformer) and with the sprint plan:
Day 4 wraps preprocessing + the AraBERT model into one `predict()` pipeline; TF-IDF remains
the interpretable baseline that Day 5's full evaluation and SHAP work will be compared against.

## 16. Key findings

1. **TF-IDF is the strongest classical representation** for this Arabic sentiment task: test
   macro-F1 = **0.8623** (accuracy 0.8623, ROC-AUC 0.9417) with a plain Logistic Regression —
   above both mean-pooled static-embedding baselines (Word2Vec-mean **0.8350**,
   AraBERT-static-mean **0.8313**) on the identical test set. The reason is structural: TF-IDF's
   idf weighting keeps word-importance information that naive mean-pooling averages away.
2. **The vocabulary choice was evidence-driven, not assumed:** a small validation sweep
   (3k / 5k / 10k / 20k) selected **10,000** features; 5,000 (the curriculum's starting point)
   was measurably worse on validation (0.8493 vs 0.8503), and 20,000 over-fitted slightly.
3. **The carried-forward Sprint-2 improvement was applied:** 5-fold stratified CV on the train
   split gives macro-F1 = **0.8556 ± 0.0033** — the pipeline is stable, and the single held-out
   test number (0.8623) agrees closely with the CV mean.
4. **Semantic geometry is real and demonstrable in both embedding spaces** — polarity and
   consumer-domain clusters appear in the domain Word2Vec space and in the pre-trained
   AraBERT static space (synonyms, verb inflections, product attributes).
5. **Contextual embeddings remain the best representation overall:** Week-7 AraBERT v2 macro-F1
   = **0.9000** on the same test set beats every classical Day-2 baseline, confirming the Week 7
   architecture decision (text → transformer) and motivating Day 4's integration work.

## 17. Limitations (documented honestly)

1. **Static embeddings evaluated with a linear head only.** Mean-pooled word vectors are the
   curriculum's practical baseline; an attentive/neural aggregator would close some of the gap
   to TF-IDF, but that is a deep-learning experiment, not a representation comparison — and the
   Week-7 contextual model already covers that route.
2. **Bag-of-words cannot model negation scope.** TF-IDF keeps negation tokens (`ليس`, `لا`) as
   features but unigrams cannot tell *what* is negated; the Day-1 protection decision still
   helps, and the contextual model (Week 7) handles scope properly. A word bigram variant of
   the TF-IDF baseline is a natural Day-5 ablation.
3. **Small-corpus Word2Vec noise.** Domain Word2Vec was trained on 14k reviews; some rare-word
   neighbourhoods are noisy (e.g. `مكسور`). Pre-trained AraBERT neighbours are cleaner. This is
   a corpus-size effect, not a method failure.
4. **AraBERT static projection is crude.** Averaging WordPiece sub-word vectors approximates a
   word embedding but loses the contextual layers — it is used here only as a *static*
   pre-trained Arabic source, not as a proxy for the fine-tuned model.
5. **No same-task LSTM comparison exists.** Week-7 Day-3's LSTM ran on an ECG signal dataset;
   it is quoted for transparency but cannot be compared on this text task.
6. **Day-1 cleaning is reused as-is.** The TF-IDF/embedding results inherit Day-1's documented
   decisions (negation-protected stop-word handling, train-fitted lemma table). We did not
   rebuild a second cleaning pipeline — that is the continuity contract of this sprint.
7. **Matplotlib does not shape Arabic script** (letters render unjoined in figures), so figures
   use Latin axis labels; Arabic terms appear in the printed tables above them.
